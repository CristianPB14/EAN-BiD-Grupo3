```python

import os
from sqlalchemy import create_engine, text

# 2. Parámetros de conexión (el host es el nombre del servicio en el YAML)
host = "db"
db_user     = os.environ["POSTGRES_USER"]
db_password = os.environ["POSTGRES_PASSWORD"]
db_name     = os.environ["POSTGRES_DB"]

# 3. Cadena de conexión
engine = create_engine(
    f"postgresql+psycopg2://{db_user}:{db_password}@{host}:5432/{db_name}"
)


with engine.connect() as conn:
    version = conn.execute(text("SELECT version();")).scalar()
print("Conexion establecida por nombre de servicio 'db'")
print("Motor:", version)
```
